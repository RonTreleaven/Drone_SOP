```python
import json
import csv
from pathlib import Path

```

```python
# adjust filenames
infile = Path("./new_stations.cache.json")
outfile = Path("./new_stations.csv")

```

```python
# load JSON (supports either a list at top-level or newline-delimited JSON)
raw = infile.read_text(encoding="utf-8").strip()
if raw.startswith("["):
    data = json.loads(raw)
else:
    # NDJSON: one JSON object per line
    data = [json.loads(line) for line in raw.splitlines() if line.strip()]

# collect all unique keys across objects (flatten only top-level fields)
keys = []
for obj in data:
    for k in obj.keys():
        if k not in keys:
            keys.append(k)

# write CSV
with outfile.open("w", newline="", encoding="utf-8") as f:
    writer = csv.DictWriter(f, fieldnames=keys, extrasaction="ignore")
    writer.writeheader()
    for obj in data:
        # ensure every row has all keys
        row = {k: obj.get(k, "") for k in keys}
        writer.writerow(row)

print(f"Wrote {len(data)} rows to {outfile}")
```
