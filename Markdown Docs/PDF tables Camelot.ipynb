```python
import camelot

tables = camelot.read_pdf("canadian-notam-operating-procedures-cnop-v11.pdf", pages="199-281", flavor="stream")

# for i, table in enumerate(tables):
#     table.df.to_csv(f"table_{i}.csv")
```

```python
print(len(tables))
```

```python
for i, table in enumerate(tables[:5]):
    print(f"\n--- Table {i} ---")
    print(table.df.head())
```

```python
valid_tables = []

for table in tables:
    df = table.df
    
    # crude filter: look for rows that contain 2-letter codes
    if df.apply(lambda row: row.astype(str).str.match(r'^[A-Z]{2}$').any(), axis=1).any():
        valid_tables.append(df)

print(len(valid_tables))
```

```python
subjects = {}

for df in valid_tables:
    for _, row in df.iterrows():
        code = row[0].strip()
        desc = row[1].strip() if len(row) > 1 else ""
        
        if len(code) == 2 and code.isalpha():
            subjects[code] = desc

print(subjects)
```

```python
conditions = {}

for df in valid_tables:
    for _, row in df.iterrows():
        code = row[0].strip()
        desc = row[1].strip() if len(row) > 1 else ""
        
        if len(code) == 2 and code.isalpha():
            conditions[code] = desc

print(conditions)
```

```python
def decode_q(qcode):
    subject = qcode[1:3]
    condition = qcode[3:5]
    
    return {
        "subject_code": subject,
        "subject": subjects.get(subject, "Unknown"),
        "condition_code": condition,
        "condition": conditions.get(condition, "Unknown")
    }

print(decode_q("QNBXX"))
```
